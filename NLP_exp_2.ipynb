```python
#select the set of pos and neg
positive_tweets = twitter_samples.strings('positive_tweets.json')
negative_tweets = twitter_samples.strings('negative_tweets.json')
print("Name:umaid ansari","UIN:231A048")
```

```python
positive_tweets=pd.DataFrame(positive_tweets, columns=["Positive "])
positive_tweets['Sentiments']='positive'
positive_tweets.tail()
print("Name:umaid ansari","UIN:231A048")
```

```python
negative_tweets=pd.DataFrame(negative_tweets, columns=['Negative Tweets'])
negative_tweets['Sentiments']='Negative'
negative_tweets.tail()
print("Name:umaid ansari","UIN:231A048")
```

```python
tweets=pd.concat([positive_tweets,negative_tweets],ignore_index= False)
tweets=tweets.sample(frac=1).reset_index(drop=True)
tweets
print("Name:umaid ansari","UIN:231A048")
```

```python
tweets[tweets['Sentiments']=='positive'].value_counts().sum()
tweets[tweets['Sentiments']=='Negative'].value_counts().sum()
print("Name:umaid ansari","UIN:231A048")
```
